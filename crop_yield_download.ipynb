```python
#Data Cleaning 
```

```python
crop_df = crop_df.dropna(subset=['YIELD_KG'])
crop_df
```

```python
crop_df['AREA_ACRES'] = crop_df['AREA_ACRES'].fillna(crop_df['AREA_ACRES'].mean())
```

```python
crop_df['DISTRICT'] = crop_df['DISTRICT'].fillna('Unknown')
```

```python
crop_df['FERTILIZER_USED'] = crop_df['FERTILIZER_USED'].fillna('NO FERTILIZER')
```

```python
crop_df['YIELD_VALUE'] = crop_df['YIELD_VALUE'].fillna(crop_df['YIELD_VALUE'].mean())
```

```python
crop_df['HARVEST_DATE'] = pd.to_datetime( crop_df['HARVEST_DATE'], format='%d/%m/%Y', errors='coerce')
```

```python
crop_df['FARM_ID'] = crop_df['FARM_ID'].str.replace('F', '', regex=False).astype(int)
```

```python
crop_df.isna().sum()
```

```python
crop_df.dtypes
```

```python
total_yield = crop_df.groupby('CROP_TYPE').agg({'YIELD_KG': 'sum'})
total_yield
```

```python
category = total_yield.index
values = total_yield['YIELD_KG']

plt.bar(category, values, color='green', width=0.5)

plt.xlabel('Crop Type')
plt.ylabel('Total Yield (KG)')
plt.title('Total Yield by Crop Type')

plt.show()
```

```python
yield_by_date = crop_df.pivot_table(values='YIELD_KG',index='HARVEST_DATE',aggfunc='sum')

yield_by_date
```

```python
yield_by_date = yield_by_date.sort_index()

yield_by_date
```

```python
x = yield_by_date.index
y = yield_by_date['YIELD_KG']

plt.plot(x, y)

plt.xlabel('Harvest Date')
plt.ylabel('Yield (KG)')
plt.title('Yield Trend Over Time')
plt.xticks(rotation=45)

plt.show()

# too many points i guess ):
```
